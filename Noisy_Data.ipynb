NLP Preprocessing: Handling Noisy Text and ASCII

```python
import re
import unicodedata

def clean_text(text):
    text = unicodedata.normalize('NFKC', text)
    text = text.lower()
    text = re.sub(r'(.)\1{2,}', r'\1\1', text)
    text = re.sub(r'[^a-zA-Z0-9\s.,!?\']', '', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

text1 = "OMG!! That was SOOOO coool!!!!  but wait... what??? @user #nlp \x00\x1f"

print("1:", clean_text(text1))
```

```python
text2 = "Thiiisss   is   terrrribleee   spelling & bad formatting... right??? $$100"
print("2:", clean_text(text2))
```

```python
text3 = "Heeeyyy guysss!! Can't beliiieve this happened ^_^ <tag> \t\n WOWWW"
print("3:", clean_text(text3))
```
