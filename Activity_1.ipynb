```python
import warnings
warnings.filterwarnings("ignore")
```

```python
import nltk
nltk.download('punkt_tab')
```

```python
from nltk.tokenize import sent_tokenize, word_tokenize

text = """Natural language processing (NLP) is a fascinating field of
artificial intelligence. It involves analyzing and understanding
human languages, such as English, to derive meaningful insights."""

print("Sentence Tokenization:")
sentences = sent_tokenize(text)
for i, sentence in enumerate(sentences, 1):
    print(f"Sentence {i}: {sentence}")
```

```python
print("\nWord Tokenization:")
for i, sentence in enumerate(sentences, 1):
    words = word_tokenize(sentence)
    print(f"Words in Sentence {i}: {words}")
```

```python
import stanza

stanza.download("en", processors="tokenize", verbose=False)
nlp = stanza.Pipeline("en", processors="tokenize", verbose=False)

text = """Dr. John Smith, who earned his Ph.D. at M.I.T.,
visited NASA. He saw the Mars rover at 2:30 p.m. on 10/11/12.
Isn't it great? Contact him at dr.john@nasa.gov for more info."""

doc = nlp(text)

print("Sentence Tokenization:")

for i, sentence in enumerate(doc.sentences, 1):
    print(f"Sentence {i}: {sentence.text.strip()}")

print("\nWord Tokenization per Sentence:")

for i, sentence in enumerate(doc.sentences, 1):
    print(f"Sentence {i}: {sentence.text.strip()}")

print("\nWord Tokenization per Sentence:")
for i, sentence in enumerate(doc.sentences, 1):
    words = [word.text for word in sentence.words]
    print(f"Words in Sentence {i}: {words}")
```

Case Study(Using NLTK)

```python
from nltk.tokenize import sent_tokenize, word_tokenize

reviews = [
    "The product quality is excellent! I am very satisfied with the purchase. Will definitely recommend it.",
    "Terrible customer service. I had to wait two weeks for a response. Not buying again!",
    "Average product. Delivery was on time, but the packaging was damaged.",
    "The setup was simple. The battery life is amazing. However, the screen scratched easily. I am returning it tomorrow."
]

pos_words = {"excellent", "satisfied", "recommend", "great", "good", "amazing", "simple"}
neg_words = {"terrible", "damaged", "bad", "scratched", "poor", "worst"}

for i, review in enumerate(reviews, 1):
    sentences = sent_tokenize(review)
    words = [w.lower() for w in word_tokenize(review) if w.isalnum()]

    found_pos = set(words) & pos_words
    found_neg = set(words) & neg_words

    print(f"\nReview #{i} ({len(sentences)} sentences, {len(words)} total words):")
    if len(sentences) > 3:
        print("  -> FLAGGED: Review is longer than 3 sentences.")
    print(f"  Positive hits: {list(found_pos) or 'none'}")
    print(f"  Negative hits: {list(found_neg) or 'none'}")
```
